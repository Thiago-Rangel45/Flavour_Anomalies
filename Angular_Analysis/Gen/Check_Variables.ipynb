```python
import ROOT

# 1. Abrir o arquivo e capturar o Workspace/Dataset
f = ROOT.TFile.Open("effDataset_GENOnly_b0.root")
ws = f.Get("ws_b0p0")
data = ws.data("data_genDen_ev_b0")

# 2. Obter as variáveis do Workspace
ctK = ws.var("ctK")
ctL = ws.var("ctL")
phi = ws.var("phi")

# 3. Criar o Canvas e dividir em 3 partes
c = ROOT.TCanvas("c1", "Bin 0 - Angular Variables", 1200, 400)
c.Divide(3, 1)

# --- Plot Cos(Theta_K) ---
c.cd(1)
frame_k = ctK.frame(ROOT.RooFit.Title("cos(#theta_{K})"))
data.plotOn(frame_k, ROOT.RooFit.MarkerStyle(20), ROOT.RooFit.MarkerSize(0.8))
frame_k.Draw()

# --- Plot Cos(Theta_L) ---
c.cd(2)
frame_l = ctL.frame(ROOT.RooFit.Title("cos(#theta_{L})"))
data.plotOn(frame_l, ROOT.RooFit.MarkerStyle(20), ROOT.RooFit.MarkerSize(0.8), ROOT.RooFit.MarkerColor(ROOT.kRed))
frame_l.Draw()

# --- Plot Phi ---
c.cd(3)
frame_phi = phi.frame(ROOT.RooFit.Title("#phi (rad)"))
data.plotOn(frame_phi, ROOT.RooFit.MarkerStyle(20), ROOT.RooFit.MarkerSize(0.8), ROOT.RooFit.MarkerColor(ROOT.kBlue))
frame_phi.Draw()

# 4. Renderizar o Canvas no Jupyter
c.Draw()
c.SaveAs("teste.png")
```
