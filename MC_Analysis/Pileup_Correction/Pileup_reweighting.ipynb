```python
import ROOT
import pandas as pd
import numpy as np
import os
```

```python
ROOT.EnableImplicitMT()
```

```python
mc_weight = np.load('mc_weight.npy')
```

```python
Data_dir = '../Data/Raw_Data/ParkingDoubleMuonLowMass0/BPH_NanoAOD_Run2022C/250916_145116/0000'
MC_dir   = '../Data/Raw_Data/MC'

chain_MC = ROOT.TChain('Events')
chain_MC.Add(MC_dir + '/BPH_NANO_124.root')
chain_MC.Add(MC_dir + '/BPH_NANO_130.root')

chain_Data = ROOT.TChain('Events')
MAX_SIZE_BYTES = 200 * 1024 * 1024 * 1024
current_mc_size = 0

print(f"Adicionando arquivos Dados do diretório: {Data_dir}")
for file_name in os.listdir(Data_dir):
    if file_name.endswith(".root") and "sys" not in file_name:
        full_path = os.path.join(Data_dir, file_name)
        file_size = os.path.getsize(full_path)
        if current_mc_size + file_size <= MAX_SIZE_BYTES:
            chain_Data.Add(full_path)
            current_mc_size += file_size
        else:
            break
```

```python
ROOT.gInterpreter.Declare(f"const std::vector<double> pileup_weights = {{{','.join(map(str, mc_weight))}}};")
ROOT.gInterpreter.Declare("""
double getPileupWeight(float nTrueInt) {
    int bin = static_cast<int>(nTrueInt + 0.5);
    if (bin >= 0 && bin < pileup_weights.size()) {
        return pileup_weights[bin];
    }
    return 1.0;
}
""")
```

```python
df_data_initial = ROOT.RDataFrame(chain_Data)
df_mc_initial = ROOT.RDataFrame(chain_MC)
```

```python
def plot_mc_variable(df, variable, cut_string, bins, x_min, x_max, x_label, y_label, title):
    canvas = ROOT.TCanvas(f"c_{variable}", "canvas", 800, 600)
    Pileup_weight = "Pileup_weight"
    df_with_cut = df.Define(Pileup_weight, "getPileupWeight(Pileup_nTrueInt)") \
                    .Define(f"{variable}_cut", f"{variable} * ({cut_string})")

    histo_model = ROOT.RDF.TH1DModel(f"h_{variable}", f"{title};{x_label};{y_label}", bins, x_min, x_max)
    histo = df_with_cut.Histo1D(histo_model, f"{variable}_cut", "Pileup_weight")

    histo_ptr = histo.GetValue()
    histo_ptr.SetLineColor(ROOT.kBlue + 2)
    histo_ptr.SetFillColor(ROOT.kAzure - 9)
    histo_ptr.SetLineWidth(2)
    histo_ptr.Draw("HIST")

    canvas.Update()
    return canvas, histo_ptr
```

```python
def plot_data_variable(df, variable, cut_string, bins, x_min, x_max, x_label, y_label, title):
    canvas = ROOT.TCanvas(f"c_{variable}", "canvas", 800, 600)
    df_with_cut = df.Define(f"{variable}_cut", f"{variable}[{cut_string}]")

    histo_model = ROOT.RDF.TH1DModel(f"h_{variable}", f"{title};{x_label};{y_label}", bins, x_min, x_max)
    histo = df_with_cut.Histo1D(histo_model, f"{variable}_cut")

    histo_ptr = histo.GetValue()
    histo_ptr.SetLineColor(ROOT.kBlue + 2)
    histo_ptr.SetFillColor(ROOT.kAzure - 9)
    histo_ptr.SetLineWidth(2)
    histo_ptr.Draw("HIST")

    canvas.Update()
    return canvas, histo_ptr
```

```python
def plot_data_mc_comparison(df_data, df_mc, variable, cut_string, bins, x_min, x_max, x_label, y_label, title):
    canvas = ROOT.TCanvas("canvas", "", 800, 600)

    df_data_cut = df_data.Define(f"{variable}_cut", f"{variable}[{cut_string}]")
    df_mc_cut = (df_mc.Define("Pileup_weight", "getPileupWeight(Pileup_nTrueInt)") 
                      .Define("Total_weight", f"Pileup_weight * {lumi_weight}")   
                      .Define(f"{variable}_cut", f"{variable}[{cut_string}]")
                  )
    histo_model = ROOT.RDF.TH1DModel(f"h_{variable}", f"{title};{x_label};{y_label}", bins, x_min, x_max)
    hist_data = df_data_cut.Histo1D(histo_model, f"{variable}_cut")
    hist_mc = df_mc_cut.Histo1D(histo_model, f"{variable}_cut", "Total_weight")

    hist_ptr_data = hist_data.GetValue()
    hist_ptr_mc   = hist_mc.GetValue()

    hist_ptr_data.Draw("PE")
        
    hist_ptr_mc.SetLineColor(ROOT.kBlue + 2)
    #hist_ptr_mc.SetFillColor(ROOT.kAzure - 9)
    hist_ptr_mc.SetLineWidth(2)
    hist_ptr_mc.Draw("HIST SAME")
    canvas.Draw()

    return canvas, hist_ptr_data, hist_ptr_mc
```
