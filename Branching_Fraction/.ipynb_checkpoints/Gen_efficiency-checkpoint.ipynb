```python
import uproot
import dask_awkward as dak
import awkward as ak
import numpy as np
import pandas as pd
import vector
from dask.diagnostics import ProgressBar
import os 

vector.register_awkward() 
file_path = "/cms/store/user/tdeandra/BPH_NanoAOD_MC/BdtoKstar2Mu_KstartoKPi_TuneCP5_13p6TeV_pythia8-evtgen/BPH_Nano_MC_BdtoKstarMuMu_NoFilter_2022/260427_132612/0000/*.root"

print("Lendo arquivos e montando o grafo Lazy...")
df = uproot.dask(
    {file_path: "Events"},
    library="ak"
) 

# ==============================================================================
# 1. TRUTH MATCHING (INALTERADO)
# ==============================================================================
def get_signal_mask(pdg, mother):
    local_idx = ak.local_index(pdg)

    safe_mom = ak.where(mother >= 0, mother, 0)
    mom_pdg  = pdg[safe_mom]
    mom_pdg  = ak.where(mother >= 0, mom_pdg, -999)

    safe_grandmom = safe_mom[safe_mom]
    grandmom_pdg  = pdg[safe_grandmom]
    grandmom_pdg  = ak.where(safe_mom >= 0, grandmom_pdg, -999)

    B_ancestor_idx = ak.full_like(pdg, -1)
    B_ancestor_idx = ak.where((pdg == 511), local_idx, B_ancestor_idx)
    B_ancestor_idx = ak.where((pdg == 313) & (mom_pdg == 511), safe_mom, B_ancestor_idx)
    B_ancestor_idx = ak.where((pdg == 22)  & (mom_pdg == 511), safe_mom, B_ancestor_idx)

    is_K    = (pdg == 321)  & (mom_pdg == 313) & (grandmom_pdg == 511)
    is_pi   = (pdg == -211) & (mom_pdg == 313) & (grandmom_pdg == 511)
    is_mu_m = (pdg == 13)   & (mom_pdg == 22)  & (grandmom_pdg == 511)
    is_mu_p = (pdg == -13)  & (mom_pdg == 22)  & (grandmom_pdg == 511)

    B_ancestor_idx = ak.where(is_K,               safe_grandmom, B_ancestor_idx)
    B_ancestor_idx = ak.where(is_pi,              safe_grandmom, B_ancestor_idx)
    B_ancestor_idx = ak.where(is_mu_m | is_mu_p,  safe_grandmom, B_ancestor_idx)

    anc_kstar = B_ancestor_idx[(pdg == 313) & (mom_pdg == 511)]
    anc_K     = B_ancestor_idx[is_K]
    anc_pi    = B_ancestor_idx[is_pi]
    anc_mu_m  = B_ancestor_idx[is_mu_m]
    anc_mu_p  = B_ancestor_idx[is_mu_p]

    p_anc = B_ancestor_idx[:, :, np.newaxis]

    has_1_kstar = ak.sum(p_anc == anc_kstar[:, np.newaxis, :], axis=-1) == 1
    has_1_K     = ak.sum(p_anc == anc_K[:, np.newaxis, :],     axis=-1) == 1
    has_1_pi    = ak.sum(p_anc == anc_pi[:, np.newaxis, :],    axis=-1) == 1
    has_1_mu_m  = ak.sum(p_anc == anc_mu_m[:, np.newaxis, :],  axis=-1) == 1
    has_1_mu_p  = ak.sum(p_anc == anc_mu_p[:, np.newaxis, :],  axis=-1) == 1

    is_valid_chain  = has_1_kstar & has_1_K & has_1_pi & has_1_mu_m & has_1_mu_p
    particle_mask   = is_valid_chain & (B_ancestor_idx >= 0)

    # Retorna APENAS o jagged array de booleanos
    return particle_mask 

# Definindo a função que faltava para aplicar a sua máscara nas colunas
def apply_truth_match(df):
    mask = get_signal_mask(df["BPHGenPart_pdgId"], df["BPHGenPart_genPartIdxMother"])
    
    df["BPHGenPart_pdgId"] = df["BPHGenPart_pdgId"][mask]
    df["BPHGenPart_pt"]    = df["BPHGenPart_pt"][mask]
    df["BPHGenPart_eta"]   = df["BPHGenPart_eta"][mask]
    df["BPHGenPart_phi"]   = df["BPHGenPart_phi"][mask]
    df["BPHGenPart_mass"]  = df["BPHGenPart_mass"][mask]
    
    return df

# Verificando as partículas B0 originais na máscara (passo opcional do seu script)
mask = get_signal_mask(df["BPHGenPart_pdgId"], df["BPHGenPart_genPartIdxMother"]) 
with ProgressBar():
    pdg_id = df["BPHGenPart_pdgId"][mask].compute() 

pdg_id = pdg_id[pdg_id == 511]
Number_B0 = ak.flatten(pdg_id)
print("Número de eventos com B0:", len(pdg_id))
print("Total de B0 flatten:", len(Number_B0)) 

# ==============================================================================
# 2. SELEÇÃO E ACEITAÇÃO
# ==============================================================================
def apply_selection(df):
    """
    Aplica cortes cinemáticos. Como aplicamos a máscara no passo anterior, 
    todos os pdgs, pts e etas já são exclusivamente do decaimento de sinal.
    """
    pdg = df["BPHGenPart_pdgId"]
    pt  = df["BPHGenPart_pt"]
    eta = df["BPHGenPart_eta"]

    # Identifica múons e hádrons
    is_mu = (abs(pdg) == 13)
    is_h  = ((abs(pdg) == 321) | (abs(pdg) == 211))

    # Aplica os cortes 
    mu_pass = is_mu & (pt > 2.0) & (abs(eta) < 2.4)
    h_pass  = is_h  & (pt > 1.5) & (abs(eta) < 2.4)

    # O evento só sobrevive se os 2 múons e os 2 hádrons passaram no corte
    event_pass = (ak.sum(mu_pass, axis=1) == 2) & (ak.sum(h_pass, axis=1) == 2)

    return df[event_pass] 

# ==============================================================================
# 3. CÁLCULO DE MASSA INVARIANTE (q^2)
# ==============================================================================
def calculate_q2(df):
    is_mu = (abs(df["BPHGenPart_pdgId"]) == 13)
    muons_all = df["BPHGenPart_pt"][is_mu]
    
    # Criamos uma máscara: apenas eventos que tenham pelo menos 2 múons
    has_two_muons = (ak.num(muons_all) >= 2)
    df = df[has_two_muons]    
    is_mu_filtered = (abs(df["BPHGenPart_pdgId"]) == 13)
    
    muons_4v = ak.zip(
        {
            "pt":   df["BPHGenPart_pt"][is_mu_filtered],
            "eta":  df["BPHGenPart_eta"][is_mu_filtered],
            "phi":  df["BPHGenPart_phi"][is_mu_filtered],
            "mass": df["BPHGenPart_mass"][is_mu_filtered]
        },
        with_name="Momentum4D"
    )

    # Agora o índice 0 e 1 existem com certeza para todos os eventos em muons_4v
    dimuon = muons_4v[:, 0] + muons_4v[:, 1]
    df["q"] = dimuon.mass
    return df

# ==============================================================================
# PIPELINE DE PROCESSAMENTO
# ==============================================================================
df = apply_truth_match(df)
#df = apply_selection(df)
df = calculate_q2(df)

features = [
    "BPHGenPart_pt",
    "BPHGenPart_eta",
    "BPHGenPart_phi",
    "BPHGenPart_mass",
    "q",
    "BPHGenPart_pdgId"
] 

with ProgressBar():
    B = df[features].compute()    

# ==============================================================================
# CONVERSÃO E ANÁLISE NO PANDAS
# ==============================================================================
df_pandas = ak.to_dataframe(B)
df_pandas = df_pandas.reset_index() 
df_pandas['q2'] = df_pandas['q']**2

bins = [1.1, 2, 4.3, 6, 8.68, 10.09, 12.86, 14.18, 16, 19]
labels = [0, 1, 2, 3, 4, 5, 6, 7, 8]

# O "astype(float)" estava na mesma linha no seu script, causando SyntaxError
df_pandas['bin_index'] = pd.cut(df_pandas['q2'], bins=bins, labels=labels, right=False).astype(float) 

is_B0 = df_pandas["BPHGenPart_pdgId"] == 511

print(f"Total de partículas no DF Gen: {len(df_pandas)}")

for i in range(9):  
    is_bin = df_pandas["bin_index"] == i    
    df_filtrado = df_pandas[is_B0 & is_bin]    
    print(f"Bin {i}: {len(df_filtrado)} partículas B0 encontradas")
```

```python
import uproot
import dask_awkward as dak
import awkward as ak
import numpy as np
import pandas as pd
import vector
from dask.diagnostics import ProgressBar
import os 

vector.register_awkward() 
file_path = "/cms/store/user/tdeandra/BPH_NanoAOD_MC/BdtoJpsiKstar_Jpsito2Mu_KstartoKPi_TuneCP5_13p6TeV_pythia8-evtgen/BPH_Nano_MC_BdtoJpsiKstar_NoFilter_2022_retry1/260427_173326/0000/*.root"

print("Lendo arquivos e montando o grafo Lazy...")
df = uproot.dask(
    {file_path: "Events"},
    library="ak"
) 

# ==============================================================================
# 1. TRUTH MATCHING (INALTERADO)
# ==============================================================================
def get_signal_mask(pdg, mother):
    
    local_idx = ak.local_index(pdg)
    safe_mom      = ak.where(mother >= 0, mother, 0)
    mom_pdg       = pdg[safe_mom]
    mom_pdg       = ak.where(mother >= 0, mom_pdg, -999)

    safe_grandmom = safe_mom[safe_mom]
    grandmom_pdg  = pdg[safe_grandmom]
    grandmom_pdg  = ak.where(safe_mom >= 0, grandmom_pdg, -999)

    safe_greatgrandmom = safe_mom[safe_grandmom]
    greatgrandmom_pdg  = pdg[safe_greatgrandmom]
    greatgrandmom_pdg  = ak.where(safe_grandmom >= 0, greatgrandmom_pdg, -999)

   
    B_ancestor_idx = ak.full_like(pdg, -1)
    B_ancestor_idx = ak.where((pdg == 511), local_idx, B_ancestor_idx)
    
    # K*0 e J/psi filhos diretos do B0
    B_ancestor_idx = ak.where((pdg == 313) & (mom_pdg == 511), safe_mom, B_ancestor_idx)
    B_ancestor_idx = ak.where((pdg == 443) & (mom_pdg == 511), safe_mom, B_ancestor_idx)

    # γ filho do J/psi, neto do B0
    B_ancestor_idx = ak.where((pdg == 22) & (mom_pdg == 443) & (grandmom_pdg == 511), safe_grandmom, B_ancestor_idx)
    
    # K+ e π-: filhos do K*0, netos do B0
    is_K  = (pdg == 321)  & (mom_pdg == 313) & (grandmom_pdg == 511)
    is_pi = (pdg == -211) & (mom_pdg == 313) & (grandmom_pdg == 511)

    B_ancestor_idx = ak.where(is_K,  safe_grandmom, B_ancestor_idx)
    B_ancestor_idx = ak.where(is_pi, safe_grandmom, B_ancestor_idx)

    is_mu_m = ((pdg == 13)  & (mom_pdg == 22) & (grandmom_pdg == 443) & (greatgrandmom_pdg == 511))
    is_mu_p = ((pdg == -13) & (mom_pdg == 22) & (grandmom_pdg == 443) & (greatgrandmom_pdg == 511))

    B_ancestor_idx = ak.where(is_mu_m | is_mu_p, safe_greatgrandmom, B_ancestor_idx)

    anc_kstar = B_ancestor_idx[(pdg == 313) & (mom_pdg == 511)]
    anc_jpsi  = B_ancestor_idx[(pdg == 443) & (mom_pdg == 511)]
    anc_K     = B_ancestor_idx[is_K]
    anc_pi    = B_ancestor_idx[is_pi]
    anc_mu_m  = B_ancestor_idx[is_mu_m]
    anc_mu_p  = B_ancestor_idx[is_mu_p]
    p_anc     = B_ancestor_idx[:, :, np.newaxis]

    has_1_kstar = ak.sum(p_anc == anc_kstar[:, np.newaxis, :], axis=-1) == 1
    has_1_jpsi  = ak.sum(p_anc == anc_jpsi[:, np.newaxis, :],  axis=-1) == 1
    has_1_K     = ak.sum(p_anc == anc_K[:, np.newaxis, :],     axis=-1) == 1
    has_1_pi    = ak.sum(p_anc == anc_pi[:, np.newaxis, :],    axis=-1) == 1
    has_1_mu_m  = ak.sum(p_anc == anc_mu_m[:, np.newaxis, :],  axis=-1) == 1
    has_1_mu_p  = ak.sum(p_anc == anc_mu_p[:, np.newaxis, :],  axis=-1) == 1

    is_valid_chain = (has_1_kstar & has_1_jpsi & has_1_K & has_1_pi & has_1_mu_m  & has_1_mu_p)
    particle_mask = is_valid_chain & (B_ancestor_idx >= 0)

    return particle_mask

# Definindo a função que faltava para aplicar a sua máscara nas colunas
def apply_truth_match(df):
    mask = get_signal_mask(df["BPHGenPart_pdgId"], df["BPHGenPart_genPartIdxMother"])
    
    df["BPHGenPart_pdgId"] = df["BPHGenPart_pdgId"][mask]
    df["BPHGenPart_pt"]    = df["BPHGenPart_pt"][mask]
    df["BPHGenPart_eta"]   = df["BPHGenPart_eta"][mask]
    df["BPHGenPart_phi"]   = df["BPHGenPart_phi"][mask]
    df["BPHGenPart_mass"]  = df["BPHGenPart_mass"][mask]
    
    return df

# Verificando as partículas B0 originais na máscara (passo opcional do seu script)
mask = get_signal_mask(df["BPHGenPart_pdgId"], df["BPHGenPart_genPartIdxMother"]) 
with ProgressBar():
    pdg_id = df["BPHGenPart_pdgId"][mask].compute() 

pdg_id = pdg_id[pdg_id == 511]
Number_B0 = ak.flatten(pdg_id)
print("Número de eventos com B0:", len(pdg_id))
print("Total de B0 flatten:", len(Number_B0)) 

# ==============================================================================
# 2. SELEÇÃO E ACEITAÇÃO
# ==============================================================================
def apply_selection(df):
    """
    Aplica cortes cinemáticos. Como aplicamos a máscara no passo anterior, 
    todos os pdgs, pts e etas já são exclusivamente do decaimento de sinal.
    """
    pdg = df["BPHGenPart_pdgId"]
    pt  = df["BPHGenPart_pt"]
    eta = df["BPHGenPart_eta"]

    # Identifica múons e hádrons
    is_mu = (abs(pdg) == 13)
    is_h  = ((abs(pdg) == 321) | (abs(pdg) == 211))

    # Aplica os cortes 
    mu_pass = is_mu & (pt > 2.0) & (abs(eta) < 2.4)
    h_pass  = is_h  & (pt > 1.5) & (abs(eta) < 2.4)

    # O evento só sobrevive se os 2 múons e os 2 hádrons passaram no corte
    event_pass = (ak.sum(mu_pass, axis=1) == 2) & (ak.sum(h_pass, axis=1) == 2)

    return df[event_pass] 

# ==============================================================================
# 3. CÁLCULO DE MASSA INVARIANTE (q^2)
# ==============================================================================
def calculate_q2(df):
    is_mu = (abs(df["BPHGenPart_pdgId"]) == 13)
    muons_all = df["BPHGenPart_pt"][is_mu]
    
    # Criamos uma máscara: apenas eventos que tenham pelo menos 2 múons
    has_two_muons = (ak.num(muons_all) >= 2)
    df = df[has_two_muons]    
    is_mu_filtered = (abs(df["BPHGenPart_pdgId"]) == 13)
    
    muons_4v = ak.zip(
        {
            "pt":   df["BPHGenPart_pt"][is_mu_filtered],
            "eta":  df["BPHGenPart_eta"][is_mu_filtered],
            "phi":  df["BPHGenPart_phi"][is_mu_filtered],
            "mass": df["BPHGenPart_mass"][is_mu_filtered]
        },
        with_name="Momentum4D"
    )

    # Agora o índice 0 e 1 existem com certeza para todos os eventos em muons_4v
    dimuon = muons_4v[:, 0] + muons_4v[:, 1]
    df["q"] = dimuon.mass
    return df

# ==============================================================================
# PIPELINE DE PROCESSAMENTO
# ==============================================================================
df = apply_truth_match(df)
#df = apply_selection(df)
df = calculate_q2(df)

features = [
    "BPHGenPart_pt",
    "BPHGenPart_eta",
    "BPHGenPart_phi",
    "BPHGenPart_mass",
    "q",
    "BPHGenPart_pdgId"
] 

with ProgressBar():
    B = df[features].compute()    

# ==============================================================================
# CONVERSÃO E ANÁLISE NO PANDAS
# ==============================================================================
df_pandas = ak.to_dataframe(B)
df_pandas = df_pandas.reset_index() 
df_pandas['q2'] = df_pandas['q']**2

bins = [1.1, 2, 4.3, 6, 8.68, 10.09, 12.86, 14.18, 16, 19]
labels = [0, 1, 2, 3, 4, 5, 6, 7, 8]

# O "astype(float)" estava na mesma linha no seu script, causando SyntaxError
df_pandas['bin_index'] = pd.cut(df_pandas['q2'], bins=bins, labels=labels, right=False).astype(float) 

is_B0 = df_pandas["BPHGenPart_pdgId"] == 511

print(f"Total de partículas Controle no DF Gen: {len(df_pandas)}")

for i in range(9):  
    is_bin = df_pandas["bin_index"] == i    
    df_filtrado = df_pandas[is_B0 & is_bin]    
    print(f"Bin {i}: {len(df_filtrado)} partículas B0 encontradas")
```
