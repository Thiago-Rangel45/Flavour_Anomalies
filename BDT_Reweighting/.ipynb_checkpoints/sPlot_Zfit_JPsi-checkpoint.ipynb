```python
import uproot
import awkward as ak
import numpy as np
import pandas as pd
import zfit
import matplotlib.pyplot as plt
import mplhep as hep
from hepstats.splot import compute_sweights
```

```python
plt.style.use(hep.style.CMS)
```

```python
MASS_RANGE = (5.0, 5.6)
JPSI_RANGE = (2.9, 3.3)

# Configuração das variáveis para plotar
features_config = [
    {
        "name": "pt", 
        "branch": "BToTrkTrkMuMu_fit_pt_good", 
        "label": r"$p_{T} (B^{0})$ [GeV]", 
        "bins": 50, 
        "range": (0, 50), 
        "log": True   # <--- Log ativado (útil para ver a cauda do espectro)
    },
    {
        "name": "eta", 
        "branch": "BToTrkTrkMuMu_fit_eta_good", 
        "label": r"$\eta (B^{0})$", 
        "bins": 50, 
        "range": (-2.5, 2.5), 
        "log": False
    },
    {
        "name": "phi", 
        "branch": "BToTrkTrkMuMu_fit_phi_good", 
        "label": r"$\phi (B^{0})$", 
        "bins": 50, 
        "range": (-3.15, 3.15), 
        "log": False
    },
    {
        "name": "svprob", 
        "branch": "BToTrkTrkMuMu_svprob_good", 
        "label": "Vertex Prob.", 
        "bins": 50, 
        "range": (0, 1), 
        "log": True   # <--- Log ativado (útil para ver diferenças de pureza)
    },
    {
        "name": "mumu_pt",                          # Nome interno para salvar o arquivo
        "branch": "MuMu_pt_good",               # Nome exato da branch na sua lista
        "label": r"$p_{T}(\mu^{+}\mu^{-})$ [GeV]",  # Label em LaTeX bonito
        "bins": 50, 
        "range": (0, 40),                           # Ajuste conforme necessário (ex: 0, 30 ou 0, 50)
        "log": True                                 # Log fica melhor para pT
    },
    {
        "name": "mumu_eta",                          # Nome interno para salvar o arquivo
        "branch": "MuMu_eta_good",               # Nome exato da branch na sua lista
        "label": r"$\eta(\mu^{+}\mu^{-})$ [GeV]",  # Label em LaTeX bonito
        "bins": 50, 
        "range": (-3.15, 3.15),                      # Ajuste conforme necessário (ex: 0, 30 ou 0, 50)
        "log": False                                 # Log fica melhor para pT
    },
    {
        "name": "mumu_phi",                          # Nome interno para salvar o arquivo
        "branch": "MuMu_phi_good",               # Nome exato da branch na sua lista
        "label": r"$\phi(\mu^{+}\mu^{-})$ [GeV]",  # Label em LaTeX bonito
        "bins": 50, 
        "range": (-3.15, 3.15),                      # Ajuste conforme necessário (ex: 0, 30 ou 0, 50)
        "log": False                                 # Log fica melhor para pT
    },
    {
        "name": "kstar_mass",                          # Nome interno para salvar o arquivo
        "branch": "BToTrkTrkMuMu_fit_ditrack_mass_Kpi_good",               # Nome exato da branch na sua lista
        "label": r"$m(\k^{+}\pi^{-})$ [GeV]",  # Label em LaTeX bonito
        "bins": 40, 
        "range": (0.7, 1.1),                      # Ajuste conforme necessário (ex: 0, 30 ou 0, 50)
        "log": False                                 # Log fica melhor para pT
    },
    
]

FILE_DATA = "/home/tandrade/Flavour_Anomalies/Data/final_data_skim.root"
FILE_MC = "/home/tandrade/Flavour_Anomalies/Data/final_mc_jpsi_skim.root"
```

```python
# ==============================================================================
# 2. LEITURA DE DADOS (CORRIGIDO E ROBUSTO)
# ==============================================================================
def load_data(filepath, is_mc=False):
    print(f">>> Lendo {'MC' if is_mc else 'DADOS'} de: {filepath}")
    
    with uproot.open(filepath) as f:
        tree = f["tree_ML"]
        
        branch_map = {
            "mass": "BToTrkTrkMuMu_fit_mass_Kpi_good",
            "mumu_mass": "MuMu_mass_good",
            "HLT_DoubleMu4_3_LowMass": "HLT_DoubleMu4_3_LowMass"
        }
        for feat in features_config:
            branch_map[feat["name"]] = feat["branch"]
            
        branches_to_load = list(branch_map.values())
        arrays = tree.arrays(branches_to_load, library="ak")

    # --- Nova Estratégia: Alinhamento Dinâmico ---
    # Usamos a branch 'mass' como referência de estrutura jagged
    reference_branch = arrays["BToTrkTrkMuMu_fit_mass_Kpi_good"]
    
    flat_data = {}
    for short_name, branch_name in branch_map.items():
        arr = arrays[branch_name]
        
        # Se a branch for 1D (como o Trigger), precisamos expandir para o formato da referência
        if arr.ndim == 1:
            # Broadcast transforma [True, False] em [[True, True], [False]] para bater com os candidatos
            arr_aligned = ak.broadcast_arrays(arr, reference_branch)[0]
            flat_arr = ak.flatten(arr_aligned, axis=1)
        else:
            # Se já for jagged, apenas achatamos
            flat_arr = ak.flatten(arr, axis=1)
            
        flat_data[short_name] = ak.to_numpy(flat_arr)

    # Verificação de sanidade
    lengths = {k: len(v) for k, v in flat_data.items()}
    if len(set(lengths.values())) > 1:
        raise ValueError(f"Desalinhamento detectado nos tamanhos das colunas: {lengths}")

    df = pd.DataFrame(flat_data)
    ref_len = len(df)
    print(f"    -> Conversão para DataFrame (Total de linhas: {ref_len})")

    # --- Aplicação de Cortes ---
    cut_jpsi = (df["mumu_mass"] > JPSI_RANGE[0]) & (df["mumu_mass"] < JPSI_RANGE[1])
    cut_mass = (df["mass"] > MASS_RANGE[0]) & (df["mass"] < MASS_RANGE[1])
    cut_trigger = (df['HLT_DoubleMu4_3_LowMass'] == 1)

    df_clean = df[cut_jpsi & cut_mass & cut_trigger].copy()
    print(f"    -> Após cortes: {len(df_clean)}")
    
    return df_clean
```

```python
try:
    df_data = load_data(FILE_DATA, is_mc=False)
    df_mc   = load_data(FILE_MC, is_mc=True)
    print(">>> Sucesso! df_data e df_mc estão prontos.")
except Exception as e:
    print(f"❌ Erro ao carregar: {e}")
```

```python
# ==============================================================================
# 3. MODELAGEM COM ZFIT (AJUSTADA)
# ==============================================================================
print(">>> Construindo modelo Zfit...")

obs = zfit.Space("mass", limits=MASS_RANGE)
data_zfit = zfit.Data.from_pandas(df_data["mass"], obs=obs)

# --- CORREÇÃO 1: Limites mais soltos para a Massa ---
# O range total é 5.0 a 5.6. Vamos dar liberdade para o pico 'mu' andar.
mu    = zfit.Parameter("mu", 5.28, 5.15, 5.40) 
sigma = zfit.Parameter("sigma", 0.025, 0.005, 0.100)
alpha = zfit.Parameter("alpha", 1.5, 0.1, 5.0)
n     = zfit.Parameter("n", 5.0, 0.1, 20.0)

# Background
tau   = zfit.Parameter("tau", -2.0, -10.0, 0.0)

# Yields
# Dica: Inicializar n_sig menor ajuda se o sinal for fraco
total_ev = len(df_data)
n_sig = zfit.Parameter("n_sig", total_ev * 0.1, 0, total_ev) 
n_bkg = zfit.Parameter("n_bkg", total_ev * 0.9, 0, total_ev)

# Construção das PDFs
signal_pdf_base = zfit.pdf.CrystalBall(mu=mu, sigma=sigma, alpha=alpha, n=n, obs=obs)
bkg_pdf_base    = zfit.pdf.Exponential(lambda_=tau, obs=obs)

signal_ext = signal_pdf_base.create_extended(n_sig)
bkg_ext    = bkg_pdf_base.create_extended(n_bkg)

model = zfit.pdf.SumPDF([signal_ext, bkg_ext])

# ==============================================================================
# 4. FIT E PLOTAGEM
# ==============================================================================
print(">>> Executando Fit...")

# Usamos ExtendedUnbinnedNLL para fits estendidos
nll = zfit.loss.ExtendedUnbinnedNLL(model, data_zfit)

# Adicionei tol um pouco maior para ajudar na convergência inicial
minimizer = zfit.minimize.Minuit(tol=1e-3) 
result = minimizer.minimize(nll)

print(f"Status do Fit: {'VÁLIDO ✅' if result.valid else 'FALHOU ❌'}")
print(result)

if result.valid:
    try:
        result.hesse() # Calcula erros
        print(">>> Erros calculados (Hesse).")
    except Exception as e:
        print(f">>> Aviso: Hesse falhou ({e}), mas o fit convergiu.")
```

```python
# ==============================================================================
# 5. VISUALIZAÇÃO (ESSENCIAL PARA DEBUG)
# ==============================================================================
def plot_fit_results(data, model, obs, title="Fit Result"):
    plt.figure(figsize=(10, 8))
    
    # 1. Plotar Dados
    # Precisamos criar um histograma "fake" para plotar os dados como pontos com erro
    counts, edges = np.histogram(data["mass"], bins=50, range=MASS_RANGE)
    centers = (edges[:-1] + edges[1:]) / 2
    yerr = np.sqrt(counts)
    
    hep.cms.label(data=True, label="Work in Progress", rlabel="13.6 TeV", lumi = 5.01)
    plt.errorbar(centers, counts, yerr=yerr, fmt='k.', label="Data")

    # 2. Plotar Modelo (PDF total)
    x_plot = np.linspace(MASS_RANGE[0], MASS_RANGE[1], 1000)
    # model.pdf retorna probabilidade normalizada. Para estendido, multiplicamos pelo yield total fitado
    # Ou usamos model.ext_pdf se disponível, mas safer é escalar manualmente:
    
    total_yield = result.params[n_sig]['value'] + result.params[n_bkg]['value']
    y_plot = model.pdf(x_plot).numpy() * total_yield * (edges[1] - edges[0]) # Escala pela largura do bin
    
    plt.plot(x_plot, y_plot, color='red', lw=2, label="Fit Total")

    # 3. Plotar Componentes (Opcional - Visualização rápida)
    # Background
    bkg_yield = result.params[n_bkg]['value']
    y_bkg = bkg_pdf_base.pdf(x_plot).numpy() * bkg_yield * (edges[1] - edges[0])
    plt.plot(x_plot, y_bkg, color='blue', linestyle='--', label="Background (Exp)")
    
    #plt.xlabel(r"$m(\mu\muK\pi)$ [GeV]") # Ajuste o label conforme sua física
    plt.xlabel(r"$m(K^{+}\pi^{-}\mu^{+}\mu^{-})$ [GeV]")
    plt.ylabel("Events / 0.012 GeV")
    plt.legend()
    #plt.title()
    plt.grid(True, alpha=0.3)
    plt.show()

# Plota independente de ter falhado ou não (para ver onde travou)
# Precisamos passar o dataframe original, não o zfit.Data, para o histograma numpy simples
plot_fit_results(df_data, model, obs)
```

```python
# ==============================================================================
# 5. CÁLCULO DE sWEIGHTS (HEPSTATS)
# ==============================================================================
print(">>> Calculando sWeights...")

# A mágica acontece aqui. A hepstats faz o mesmo que a RooStats::SPlot
weights = compute_sweights(model, data_zfit)

# O output é um dicionário { parametro_yield : array_de_pesos }
s_weights_sig = weights[n_sig]

# Adicionamos os pesos ao dataframe original
df_data["sweight_sig"] = s_weights_sig
```

```python
# Configuração global de estilo
plt.style.use(hep.style.CMS)

print(">>> Gerando plots comparativos com Ratio...")

for feat in features_config:
    name = feat["name"]
    print(f"   Plotting {name}...")
    
    # -----------------------------------------------------------
    # 1. Configuração dos Eixos (Main + Ratio)
    # -----------------------------------------------------------
    # height_ratios=[3, 1] define que o plot de cima é 3x maior que o de baixo
    # sharex=True remove os labels do eixo X do gráfico de cima
    fig, (ax_main, ax_ratio) = plt.subplots(
        nrows=2, 
        ncols=1, 
        figsize=(10, 12), 
        sharex=True, 
        gridspec_kw={"height_ratios": [3, 1], "hspace": 0.08}
    )

    # -----------------------------------------------------------
    # 2. Cálculos dos Histogramas
    # -----------------------------------------------------------
    # Histograma do MC
    mc_counts, bins = np.histogram(df_mc[name], bins=feat["bins"], range=feat["range"])
    
    # Histograma dos Dados (sWeights)
    data_counts, _ = np.histogram(df_data[name], bins=bins, weights=df_data["sweight_sig"])
    
    # Erro estatístico dos dados (SumW2)
    data_errs = np.sqrt(np.histogram(df_data[name], bins=bins, weights=df_data["sweight_sig"]**2)[0])
    
    # Normalização do MC para a área dos Dados
    scale_factor = np.sum(data_counts) / np.sum(mc_counts)
    mc_counts_scaled = mc_counts * scale_factor

    # Centros dos bins (necessário para plotar pontos da razão manualmente se não usar hep.histplot)
    bin_centers = (bins[:-1] + bins[1:]) / 2

    # -----------------------------------------------------------
    # 3. Plot Principal (ax_main)
    # -----------------------------------------------------------
    
    # Rótulo CMS
    hep.cms.label(
        "Work in Progress", 
        data=True, 
        lumi=5.01, 
        com=13.6, 
        loc=0, 
        ax=ax_main
    )
    
    # Desenhar MC
    hep.histplot(mc_counts_scaled, bins=bins, yerr=False, 
                 label=r"MC (Signal)", color="red", alpha=0.3, ax=ax_main, histtype="fill")
    
    # Desenhar Dados
    hep.histplot(data_counts, bins=bins, yerr=data_errs, 
                 label=r"Data ($_{s}Plot$)", color="black", histtype="errorbar", 
                 marker="o", markersize=4, elinewidth=1, ax=ax_main)
    
    ax_main.set_ylabel("Events (Weighted)")
    ax_main.legend(loc="upper right")

    # Lógica Log Scale (apenas no plot principal)
    if feat.get("log", False):
        ax_main.set_yscale("log")
        min_positive = np.min(data_counts[data_counts > 0]) if np.any(data_counts > 0) else 1
        # Ajustes de limites para log
        ax_main.set_ylim(bottom=max(0.1, min_positive * 0.1), top=np.max(data_counts)*50)
    else:
        # Ajustes de limites para linear
        ax_main.set_ylim(bottom=0, top=np.max(data_counts)*1.4)

    # -----------------------------------------------------------
    # 4. Plot de Razão (ax_ratio)
    # -----------------------------------------------------------
    
    # Cálculo da razão: Data / MC
    # np.divide com 'out' evita erros de divisão por zero
    ratio = np.divide(data_counts, mc_counts_scaled, out=np.zeros_like(data_counts), where=mc_counts_scaled!=0)
    
    # Propagação de erro para a razão (assumindo erro do MC desprezível ou apenas erro estatístico dos dados)
    # sigma_ratio = sigma_data / MC_scaled
    ratio_err = np.divide(data_errs, mc_counts_scaled, out=np.zeros_like(data_errs), where=mc_counts_scaled!=0)

    # Plotar a razão
    # Usamos errorbar padrão do matplotlib para ter controle total, ou hep.histplot
    ax_ratio.errorbar(bin_centers, ratio, yerr=ratio_err, 
                      fmt='o', color="black", markersize=4, elinewidth=1)
    
    # Linha de referência em 1.0
    ax_ratio.axhline(1.0, color='gray', linestyle='--', linewidth=1)
    
    # Configuração do eixo da razão
    ax_ratio.set_ylabel("Data / MC")
    ax_ratio.set_xlabel(feat["label"]) # O Label X fica no gráfico de baixo
    #ax_ratio.set_ylim(0.5, 1.5) # Ajuste esse range conforme a qualidade do fit (ex: 0.8 a 1.2 ou 0 a 2)
    ax_ratio.set_xlim(feat["range"])
    
    # -----------------------------------------------------------
    # 5. Salvar
    # -----------------------------------------------------------
    plt.savefig(f"Plots/sPlot_Comp_{name}_zfit.png", bbox_inches='tight')
    plt.close()

print(">>> Processo concluído com sucesso!")
```
